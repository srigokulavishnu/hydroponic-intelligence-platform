```python
data.columns = (
    data.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

data = data.rename(columns={
    "p_h": "ph",
    "electrical_conductivity": "ec"
})

print(data.columns.tolist())
```

```python
data["timestamp"] = pd.to_datetime(
    data["timestamp"],
    errors="coerce"
)

numeric_cols = [
    "temperature",
    "humidity",
    "light",
    "ph",
    "ec"
]

for col in numeric_cols:
    data[col] = pd.to_numeric(
        data[col],
        errors="coerce"
    )

before = len(data)

data = data.dropna(
    subset=["timestamp"] + numeric_cols
).copy()

data = data.sort_values("timestamp").reset_index(drop=True)

print("Rows before:", before)
print("Rows after :", len(data))
print("Removed    :", before - len(data))
```

```python
display(data[numeric_cols].describe().T)
```

```python
data = data.sort_values("timestamp").reset_index(drop=True)

start_time = data["timestamp"].min()
end_time = data["timestamp"].max()

total_duration = end_time - start_time

data["elapsed_fraction"] = (
    data["timestamp"] - start_time
) / total_duration

def assign_growth_stage(fraction):
    if fraction < 1/3:
        return "EARLY"
    elif fraction < 2/3:
        return "MIDDLE"
    else:
        return "LATE"

data["growth_stage_proxy"] = data["elapsed_fraction"].apply(
    assign_growth_stage
)

print(data["growth_stage_proxy"].value_counts())
```

```python
print("Displaying the head of the combined 'data' DataFrame:")
display(data.head())

print("\nInformation about the combined 'data' DataFrame:")
data.info()
```

```python
stage_mapping = {
    "EARLY": 0,
    "MIDDLE": 1,
    "LATE": 2
}

data["growth_stage_encoded"] = (
    data["growth_stage_proxy"]
    .map(stage_mapping)
)

print(
    data[
        [
            "timestamp",
            "growth_stage_proxy",
            "growth_stage_encoded"
        ]
    ].head(10)
)
```

```python
STAGE_RULES = {
    "EARLY": {
        "ph_low": 5.8,
        "ph_high": 6.5,
        "ec_low": 0.9,
        "ec_high": 1.3
    },
    "MIDDLE": {
        "ph_low": 5.8,
        "ph_high": 6.5,
        "ec_low": 1.2,
        "ec_high": 1.6
    },
    "LATE": {
        "ph_low": 5.8,
        "ph_high": 6.5,
        "ec_low": 1.5,
        "ec_high": 2.0
    }
}
```

```python
def create_stage_aware_label(row):
    stage = row["growth_stage_proxy"]
    rules = STAGE_RULES[stage]

    ph_ok = rules["ph_low"] <= row["ph"] <= rules["ph_high"]
    ec_ok = rules["ec_low"] <= row["ec"] <= rules["ec_high"]

    return "OPTIMAL" if (ph_ok and ec_ok) else "NON_OPTIMAL"


data["condition_label"] = data.apply(
    create_stage_aware_label,
    axis=1
)
```

```python
print("Overall:")
print(data["condition_label"].value_counts())

print("\nBy growth stage:")
print(
    pd.crosstab(
        data["growth_stage_proxy"],
        data["condition_label"]
    )
)
```

```python
from xgboost import XGBClassifier
import os
import json

# ============================================================
# SAVE DIRECTORY
# ============================================================

MODEL_DIR = (
    "/content/drive/MyDrive/"
    "Hydroponic_AI/models/nutrient/stage_aware_baseline"
)

os.makedirs(MODEL_DIR, exist_ok=True)

# ============================================================
# MODEL
# ============================================================

model = XGBClassifier(
    n_estimators=300,
    max_depth=5,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=42,
    n_jobs=-1
)

# ============================================================
# TRAIN
# ============================================================

model.fit(
    X_train,
    y_train
)

print("✅ Stage-aware XGBoost baseline trained")

# ============================================================
# SAVE MODEL
# ============================================================

MODEL_PATH = os.path.join(
    MODEL_DIR,
    "nutrient_xgboost_stage_aware_baseline.json"
)

model.save_model(MODEL_PATH)

print("✅ Model saved:")
print(MODEL_PATH)

# ============================================================
# SAVE METADATA
# ============================================================

metadata = {
    "model_name": "Nutrient XGBoost Stage-Aware Baseline",
    "model_type": "XGBClassifier",
    "task": "Binary classification",
    "target": "condition_label",

    "classes": {
        "0": "NON_OPTIMAL",
        "1": "OPTIMAL"
    },

    "features": FEATURES,

    "growth_stage_feature": "growth_stage_encoded",

    "growth_stage_mapping": {
        "EARLY": 0,
        "MIDDLE": 1,
        "LATE": 2
    },

    "label_type": "rule_derived_baseline",

    "stage_rules": STAGE_RULES,

    "training_samples": int(len(X_train)),
    "testing_samples": int(len(X_test)),

    "hyperparameters": {
        "n_estimators": 300,
        "max_depth": 5,
        "learning_rate": 0.05,
        "subsample": 0.8,
        "colsample_bytree": 0.8,
        "objective": "binary:logistic",
        "eval_metric": "logloss",
        "random_state": 42
    }
}

METADATA_PATH = os.path.join(
    MODEL_DIR,
    "model_metadata.json"
)

with open(METADATA_PATH, "w") as f:
    json.dump(metadata, f, indent=4)

print("✅ Metadata saved:")
print(METADATA_PATH)
```

```python
# ============================================================
# NUTRIENT XGBOOST - TEST METRICS
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

# ------------------------------------------------------------
# 1. Model predictions
# ------------------------------------------------------------

y_pred = model.predict(X_test)

# ------------------------------------------------------------
# 2. Calculate metrics
# ------------------------------------------------------------

accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test,
    y_pred,
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    zero_division=0
)

# ------------------------------------------------------------
# 3. Print results
# ------------------------------------------------------------

print("=" * 50)
print("NUTRIENT XGBOOST STAGE-AWARE BASELINE")
print("=" * 50)

print(f"Accuracy : {accuracy:.4f} ({accuracy * 100:.2f}%)")
print(f"Precision: {precision:.4f} ({precision * 100:.2f}%)")
print(f"Recall   : {recall:.4f} ({recall * 100:.2f}%)")
print(f"F1 Score : {f1:.4f} ({f1 * 100:.2f}%)")

# ------------------------------------------------------------
# 4. Classification report
# ------------------------------------------------------------

report = classification_report(
    y_test,
    y_pred,
    target_names=[
        "NON_OPTIMAL",
        "OPTIMAL"
    ],
    zero_division=0
)

print("\nClassification Report:")
print(report)

# ------------------------------------------------------------
# 5. Save metrics as JSON
# ------------------------------------------------------------

RESULTS_PATH = os.path.join(
    MODEL_DIR,
    "test_results.json"
)

results = {
    "model": "Nutrient XGBoost Stage-Aware Baseline",

    "dataset": "Figshare IoT Sensor Dataset",

    "task": "Binary classification",

    "target": "condition_label",

    "target_classes": {
        "0": "NON_OPTIMAL",
        "1": "OPTIMAL"
    },

    "label_type": "rule_derived_baseline",

    "metrics": {
        "accuracy": float(accuracy),
        "precision": float(precision),
        "recall": float(recall),
        "f1_score": float(f1)
    },

    "sample_counts": {
        "train": int(len(X_train)),
        "test": int(len(X_test))
    }
}

with open(RESULTS_PATH, "w") as f:
    json.dump(results, f, indent=4)

print("\n✅ Results saved:")
print(RESULTS_PATH)

# ------------------------------------------------------------
# 6. Save classification report
# ------------------------------------------------------------

REPORT_PATH = os.path.join(
    MODEL_DIR,
    "classification_report.txt"
)

with open(REPORT_PATH, "w") as f:

    f.write("NUTRIENT XGBOOST STAGE-AWARE BASELINE\n")
    f.write("=" * 50 + "\n\n")

    f.write(f"Accuracy : {accuracy:.4f} ({accuracy * 100:.2f}%)\n")
    f.write(f"Precision: {precision:.4f} ({precision * 100:.2f}%)\n")
    f.write(f"Recall   : {recall:.4f} ({recall * 100:.2f}%)\n")
    f.write(f"F1 Score : {f1:.4f} ({f1 * 100:.2f}%)\n\n")

    f.write("Classification Report\n")
    f.write("-" * 50 + "\n")
    f.write(report)

print("✅ Classification report saved:")
print(REPORT_PATH)

# ------------------------------------------------------------
# 7. Confusion matrix
# ------------------------------------------------------------

cm = confusion_matrix(y_test, y_pred)

fig, ax = plt.subplots(figsize=(6, 5))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=[
        "NON_OPTIMAL",
        "OPTIMAL"
    ]
)

disp.plot(ax=ax)

ax.set_title(
    "XGBoost – Nutrient/System Condition Baseline"
)

plt.tight_layout()

CM_PATH = os.path.join(
    MODEL_DIR,
    "confusion_matrix.png"
)

plt.savefig(
    CM_PATH,
    dpi=200,
    bbox_inches="tight"
)

plt.show()

print("✅ Confusion matrix saved:")
print(CM_PATH)
```
